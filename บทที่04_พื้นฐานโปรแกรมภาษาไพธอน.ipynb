## 4.1 ไวยากรณ์ของภาษาไพธอน

ภาษาพูดทุกภาษามีกฎในการเรียงคำและประโยค เรียกว่า **ไวยากรณ์ภาษา (Language Syntax)** ภาษาโปรแกรมก็เช่นกัน โดยไวยากรณ์กำหนดวิธีประกาศฟังก์ชัน คำสั่ง และข้อมูลต่าง ๆ

> 📌 **ไวยากรณ์ (Syntax)** ในทางวิทยาการคอมพิวเตอร์ หมายถึง **กฎที่ระบุลำดับของสัญลักษณ์ที่ถูกต้อง** เพื่อเขียนโปรแกรมให้มีโครงสร้างถูกต้องตามภาษาที่กำหนด

นักพัฒนาสื่อสารกับคอมพิวเตอร์ผ่าน 3 องค์ประกอบของภาษาโปรแกรม

| องค์ประกอบ | ความหมาย | ตัวอย่าง |
|---|---|---|
| **ไวยากรณ์ (Syntax)** | รูปแบบการเขียนถูกต้องตามกฎหรือไม่ | `print("Hi")` ถูก แต่ `print("Hi"` ผิด (ขาดวงเล็บปิด) |
| **ความหมาย (Semantics)** | คำสั่งนั้นทำอะไร มีความหมายถูกต้องหรือไม่ | `print(10 / 0)` ไวยากรณ์ถูกต้อง แต่ความหมายผิด (หารด้วยศูนย์ไม่ได้) |
| **หลักภาษา (Grammar)** | ชุดกฎทั้งหมดที่นิยามภาษานั้น | กฎของไพธอนทั้งหมดใน Python Language Reference |

ภาษาโปรแกรมหลายภาษามีไวยากรณ์คล้ายกัน เช่น C กับ Java ขณะที่บางภาษามีรูปแบบเฉพาะตัว เช่น Perl

**จุดเด่นของไพธอน** คือไวยากรณ์ **อ่านได้เหมือนภาษาอังกฤษ** ไม่ต้องจำกฎยิบย่อยมาก ผู้เริ่มต้นจึงมุ่งเรียนรู้แนวคิดการเขียนโปรแกรมได้ทันที

````python
# Syntax ถูก และ Semantics ถูก
print("ผลหาร =", 10 / 2)
````

````python
# Syntax ถูก แต่ Semantics ผิด -> เกิดข้อผิดพลาดขณะรัน (ZeroDivisionError)
print("ผลหาร =", 10 / 0)
````

````python
# Syntax ผิด -> ไพธอนไม่ยอมรันเลย (SyntaxError)
print("ผลหาร =", 10 / 2
````

### 4.1.1 ไอเด็นติฟายเออร์ (Identifiers)
> 📌 **Identifier** คือ **ชื่อ** ที่ใช้ระบุตัวแปร ฟังก์ชัน คลาส โมดูล หรือวัตถุอื่น ๆ

**กฎการตั้งชื่อ (ถ้าผิดกฎ ไพธอนจะแจ้ง SyntaxError)**
1. ต้อง **ขึ้นต้นด้วยตัวอักษร** (A–Z, a–z) หรือ **เครื่องหมายขีดล่าง (`_`)**
2. ตัวถัดไปเป็นตัวอักษร ตัวเลข 0–9 หรือ `_` ก็ได้
3. **ห้ามขึ้นต้นด้วยตัวเลข** และ **ห้ามมีอักขระพิเศษ** เช่น `@`, `$`, `%`, `-` หรือช่องว่าง
4. **ตัวพิมพ์เล็กและใหญ่ถือว่าต่างกัน (Case-sensitive)** เช่น `score`, `Score`, `SCORE` เป็นคนละชื่อกัน
5. **ห้ามใช้คำสงวน (Keywords)** เป็นชื่อ (ดูหัวข้อ 4.1.2)

| ชื่อ | ถูกหรือผิด | เหตุผล |
|---|---|---|
| `student_name` | ✅ | ตัวอักษรและ `_` |
| `_count` | ✅ | ขึ้นต้นด้วย `_` ได้ |
| `score2` | ✅ | มีตัวเลขได้ ถ้าไม่ได้อยู่หน้าสุด |
| `2score` | ❌ | ขึ้นต้นด้วยตัวเลข |
| `my-name` | ❌ | มีเครื่องหมาย `-` |
| `total$` | ❌ | มีอักขระพิเศษ `$` |
| `first name` | ❌ | มีช่องว่าง |
| `class` | ❌ | เป็นคำสงวน |

**ธรรมเนียมการตั้งชื่อ (Naming Conventions)** ไม่ผิดกฎถ้าไม่ทำตาม แต่นักพัฒนาไพธอนยึดถือร่วมกัน (ตามแนวทาง PEP 8)

| รูปแบบชื่อ | ความหมาย | ตัวอย่าง |
|---|---|---|
| ขึ้นต้นด้วยตัวพิมพ์ใหญ่ทุกคำ (CapWords) | **ชื่อคลาส** | `Student`, `BankAccount` |
| ตัวพิมพ์เล็กคั่นด้วย `_` (snake_case) | ชื่อตัวแปรและฟังก์ชัน | `total_score`, `calc_pay()` |
| ตัวพิมพ์ใหญ่ทั้งหมด | ค่าคงที่ | `PI`, `MAX_SIZE` |
| `_name` (ขีดล่างนำหน้า 1 ตัว) | **ส่วนตัว (Private)** ใช้ภายในเท่านั้น (เป็นธรรมเนียม ไม่ได้บังคับ) | `_balance` |
| `__name` (ขีดล่างนำหน้า 2 ตัว) | **ส่วนตัวแบบเข้มงวด (Strongly Private)** ไพธอนจะเปลี่ยนชื่อภายใน (name mangling) ให้เข้าถึงจากภายนอกได้ยาก | `__password` |
| `__name__` (ขีดล่างหน้าและหลัง 2 ตัว) | **ชื่อพิเศษที่ภาษากำหนด (Language-defined Special Name)** ไม่ควรตั้งเอง | `__init__`, `__name__` |

> 💡 ไพธอน 3 อนุญาตให้ใช้ **ตัวอักษร Unicode** เช่น ภาษาไทย เป็นชื่อได้ (`คะแนน = 90`) แต่ในงานจริงควรใช้ภาษาอังกฤษ เพื่อให้คนทั่วโลกอ่านโค้ดได้และไม่มีปัญหากับเครื่องมือบางตัว

### 🔬 ทดลอง: ตรวจสอบชื่อด้วยไพธอน
เมธอด `str.isidentifier()` ตรวจว่าเป็นชื่อที่ถูกรูปแบบหรือไม่ และ `keyword.iskeyword()` ตรวจว่าเป็นคำสงวนหรือไม่ ลองเพิ่มชื่ออื่นลงในลิสต์แล้วรันใหม่

````python
import keyword

def check_name(name):
    if keyword.iskeyword(name):
        return "❌ เป็นคำสงวน"
    if name.isidentifier():
        return "✅ ใช้ได้"
    return "❌ ผิดกฎการตั้งชื่อ"

names = ["student_name", "_count", "score2", "2score", "my-name", "total$",
         "first name", "class", "Class", "คะแนน", "print"]
for n in names:
    print(f"{n!r:16} {check_name(n)}")
````

> 🤔 สังเกตว่า `print` "ใช้ได้" เพราะไม่ใช่คำสงวนในไพธอน 3 (เป็นชื่อฟังก์ชัน) แต่ **ไม่ควรตั้งชื่อซ้ำกับฟังก์ชันที่มีอยู่แล้ว** เพราะจะทับฟังก์ชันเดิมจนเรียกใช้ไม่ได้ ลองรันเซลล์ด้านล่างดู

````python
print = "สวัสดี"       # ❌ ตั้งชื่อตัวแปรทับฟังก์ชัน print
try:
    print("Hello")
except TypeError as e:
    import builtins
    builtins.print("เกิดข้อผิดพลาด:", e)
finally:
    del print           # ลบตัวแปรที่ตั้งทับ เพื่อให้ print กลับมาใช้ได้ตามปกติ

print("print กลับมาใช้งานได้แล้ว ✅")
````

### 4.1.2 คำสงวน (Reserved Words)
**คำหลัก (Keywords)** คือคำที่ไพธอนสงวนไว้ใช้เป็นคำสั่งของภาษา **ห้ามนำมาตั้งเป็นชื่อ** ตัวแปร ค่าคงที่ หรือ Identifier อื่น ๆ วิธีที่แม่นยำที่สุดในการดูรายการคำสงวนคือถามจากไพธอนโดยตรง

````python
import keyword
print("จำนวนคำสงวน:", len(keyword.kwlist), "คำ\n")
for i, kw in enumerate(keyword.kwlist, 1):
    print(f"{kw:<10}", end="\n" if i % 6 == 0 else "")
````

### 4.1.3 บรรทัดและการเยื้องหน้า (Lines and Indentation)
ไพธอน **ไม่ใช้วงเล็บปีกกา `{ }`** บอกขอบเขตของบล็อกโค้ดเหมือนภาษา C หรือ Java แต่ใช้ **เครื่องหมายโคลอน (`:`) + การขึ้นบรรทัดใหม่ + การเยื้องหน้า (Indentation)** ซึ่งไพธอน **บังคับใช้อย่างเข้มงวด**

> ⚠️ ทุกบรรทัดในบล็อกเดียวกัน **ต้องเยื้องด้วยจำนวนช่องว่างเท่ากัน** โดยมาตรฐานคือ **4 ช่องว่าง** (กด Tab ใน Jupyter/PyCharm/VS Code จะได้ 4 ช่องว่างให้อัตโนมัติ)

**เปรียบเทียบกับภาษา C**
```c
if (x > 0) {
    printf("positive");
}
```
**ภาษาไพธอน**
```python
if x > 0:
    print("positive")
```

````python
if True:
    print("Answer")     # บล็อกของ if (เยื้อง 4 ช่อง)
    print("True")
else:
    print("Answer")     # บล็อกของ else
    print("False")
````

บล็อกซ้อนกันได้ โดยเพิ่มการเยื้องทีละระดับ ตัวอย่างนี้วนรับตัวเลขแล้วบอกว่าเป็นเลขคู่หรือคี่ จนกว่าจะป้อนค่าที่ ≤ 0 (คำสั่ง `while`, `if` จะได้เรียนละเอียดในบทที่ 7)

```
while True:                  ← ระดับ 0
    value = ...              ← ระดับ 1 (อยู่ในบล็อก while)
    if value <= 0:           ← ระดับ 1
        break                ← ระดับ 2 (อยู่ในบล็อก if)
```

````python
while True:
    value = int(input("Enter number (0 = stop): "))
    if value <= 0:
        break
    elif value % 2 == 0:
        print("{} is even number.".format(value))
    else:
        print("{} is odd number.".format(value))
print("จบโปรแกรม")
````

### 4.1.4 การแสดงคำสั่งหลายบรรทัด (Multi-Line Statements)
ปกติคำสั่งไพธอนจบที่การขึ้นบรรทัดใหม่ ถ้าคำสั่งยาวมาก ใช้ **เครื่องหมายแบ็กสแลช (`\`)** ปิดท้ายบรรทัด เพื่อบอกว่าบรรทัดถัดไปเป็นคำสั่งเดียวกัน

````python
item_one, item_two, item_three = 100, 250, 75
total = item_one + \
        item_two + \
        item_three
print("total =", total)
````

ข้อความที่อยู่ภายในวงเล็บ `[ ]`, `{ }` หรือ `( )` **ไม่ต้องใช้ `\`** เพราะไพธอนรู้ว่ายังไม่ปิดวงเล็บ (วิธีนี้นิยมกว่าและอ่านง่ายกว่า)

````python
days = ['Monday',
        'Tuesday',
        'Wednesday',
        'Thursday',
        'Friday']
print(days)

total = (item_one +       # ใช้วงเล็บแทน \
         item_two +
         item_three)
print("total =", total)
````

### 4.1.5 การใช้เครื่องหมายคำพูด (Quotation in Python)
ไพธอนใช้เครื่องหมายคำพูดได้ 3 แบบในการสร้าง **สตริง (string)** หรือข้อความ โดย **ต้องเปิดและปิดด้วยเครื่องหมายชนิดเดียวกัน**

| แบบ | ตัวอย่าง | ใช้เมื่อ |
|---|---|---|

### 4.2.2 การแสดงผล
คำสั่งพื้นฐานในการแสดงผลคือฟังก์ชัน `print()` ซึ่งแสดงได้หลายค่าโดยคั่นด้วยจุลภาค และมีพารามิเตอร์ที่ควรรู้ 2 ตัว คือ `sep` (ตัวคั่นระหว่างค่า ค่าเริ่มต้นคือช่องว่าง) และ `end` (ตัวปิดท้าย ค่าเริ่มต้นคือการขึ้นบรรทัดใหม่)

````python
print("ไพธอน", 3, "สนุก")               # คั่นด้วยช่องว่างอัตโนมัติ
print("2026", "10", "18", sep="-")      # เปลี่ยนตัวคั่นเป็น -
print("ไม่ขึ้นบรรทัดใหม่", end=" | ")
print("ต่อบรรทัดเดิม")
````

#### การจัดรูปแบบด้วยเมธอด `format()`
ไพธอน 3 มีเมธอด **`format()`** ของสตริง ซึ่งช่วย **แทนค่าตัวแปรลงในข้อความ** และ **จัดรูปแบบการแสดงผล** ทำให้โค้ดอ่านและใช้งานง่ายขึ้น

**Formatters** ทำงานโดยใส่ **ตัวยึดตำแหน่ง (Placeholder)** ซึ่งคือ **วงเล็บปีกกา `{}`** ลงในสตริง แล้วเรียก `format()` เพื่อนำค่าไปใส่แทนที่ตำแหน่งนั้น

````python
print("Sammy has {} balloons.".format(5))
````

กำหนดสตริงที่มีตัวยึดตำแหน่งไว้ในตัวแปรก่อนก็ได้

````python
open_string = "Sammy loves {}."
print(open_string.format("open source"))
````

#### ตัวยึดตำแหน่งหลายตัว
ใส่ `{}` หลายคู่ แล้วส่งค่าเข้า `format()` โดยคั่นด้วยจุลภาค (`,`) ค่าจะถูกใส่ **ตามลำดับ** และผสมชนิดข้อมูลได้ (สตริงกับจำนวนเต็ม)

````python
new_open_string = "Sammy loves {} {}."
print(new_open_string.format("open-source", "software"))

string = "Sammy loves {} {}, and has {} {}."
print(string.format("open-source", "software", 5, "balloons"))
````

#### จัดลำดับด้วยอาร์กิวเมนต์ตำแหน่ง (Positional) และคำหลัก (Keyword)
- `{}` ว่าง ๆ ใส่ค่าตามลำดับ
- `{0}`, `{1}`, … ระบุ **หมายเลขดัชนี** ของค่าใน `format()` ซึ่ง **เริ่มนับจาก 0** จึงสลับลำดับหรือใช้ค่าเดิมซ้ำได้
- `{ชื่อ}` ระบุด้วย **ชื่อคำหลัก** เช่น `{pr}` คู่กับ `pr="..."`

````python
print("Sammy the {} has a pet {}!".format("shark", "pilot fish"))
print("Sammy the {0} has a pet {1}!".format("shark", "pilot fish"))
print("Sammy the {1} has a pet {0}!".format("shark", "pilot fish"))     # สลับลำดับ
print("Sammy is a {3}, {2}, and {1} {0}!".format("happy", "smiling", "blue", "shark"))
````

````python
print("Sammy the {0} {1} a {pr}.".format("shark", "made", pr="pull request"))
print("Sammy the {pr} {1} a {0}.".format("shark", "made", pr="pull request"))
print("{0}, {0}, {0} ซ้ำได้".format("ไชโย"))                             # ใช้ค่าเดิมซ้ำ
````

````python
# ❌ อ้างดัชนีที่ไม่มีอยู่ -> IndexError
print("Sammy the {0} has a pet {2}!".format("shark", "pilot fish"))
````

#### การแสดงผลโดยใช้ตัวแปลงข้อมูล (Conversion)
ใช้รูปแบบ **`{field_name:conversion}`** โดย `field_name` คือดัชนีหรือชื่อของอาร์กิวเมนต์ และ `conversion` คือรหัสรูปแบบข้อมูล

| รหัส | ความหมาย | ตัวอย่าง | ผลลัพธ์ |
|---|---|---|---|
| `s` | สตริง | `"{:s}".format("hi")` | `hi` |
| `d` | จำนวนเต็มฐานสิบ | `"{:d}".format(42)` | `42` |
| `f` | ทศนิยม (ค่าเริ่มต้น 6 ตำแหน่ง) | `"{:f}".format(75)` | `75.000000` |
| `.nf` | ทศนิยม n ตำแหน่ง (ปัดเศษ) | `"{:.3f}".format(75.765367)` | `75.765` |
| `,` | คั่นหลักพัน | `"{:,}".format(1234567)` | `1,234,567` |
| `%` | ร้อยละ | `"{:.1%}".format(0.256)` | `25.6%` |
| `b` / `x` | เลขฐานสอง / ฐานสิบหก | `"{:b}".format(10)` | `1010` |

````python
print("Sammy ate {0:f} percent of a {1}!".format(75, "pizza"))
print("Sammy ate {0:.3f} percent of a pizza!".format(75.765367))
print("ราคา {:,.2f} บาท".format(1234567.891))
print("อัตราผ่าน {:.1%}".format(0.256))
print("10 ในฐานสอง = {:b}".format(10))      # ทบทวนบทที่ 1
````

#### การกำหนดขนาดของฟิลด์และการจัดตำแหน่ง
ใส่ **ตัวเลขความกว้าง** หลังเครื่องหมาย `:` เพื่อกำหนดจำนวนตัวอักษรของฟิลด์ โดยค่าเริ่มต้น **สตริงชิดซ้าย** และ **ตัวเลขชิดขวา** ส่วนการจัดตำแหน่งเปลี่ยนได้ด้วยสัญลักษณ์

| สัญลักษณ์ | การจัดตำแหน่ง |
|---|---|
| `<` | ชิดซ้าย |
| `^` | กึ่งกลาง |
| `>` | ชิดขวา |

ใส่ **อักขระเติม (fill)** ไว้หน้าสัญลักษณ์จัดตำแหน่งได้ เช่น `{:*^20}` คือเติม `*` และจัดกึ่งกลางในฟิลด์กว้าง 20 ตัวอักษร

> 💡 เซลล์ด้านล่างใส่ `|` ไว้รอบฟิลด์ เพื่อให้มองเห็นขอบเขตของแต่ละฟิลด์ได้ชัด

````python
print("Sammy has |{0:4}| red |{1:16}|!".format(5, "balloons"))     # ค่าเริ่มต้น
print("Sammy has |{0:<4}| red |{1:^16}|!".format(5, "balloons"))   # เปลี่ยนการจัดตำแหน่ง
print("{:*^20s}".format("Sammy"))                                   # เติม * กึ่งกลาง
print("{:0>5d}".format(42))                                         # เติม 0 ด้านหน้า
````

#### 🔬 ตัวอย่างประยุกต์: พิมพ์ใบเสร็จให้ตรงคอลัมน์

````python
items = [("กาแฟเย็น", 2, 45), ("ครัวซองต์", 1, 65.5), ("น้ำเปล่า", 3, 10)]

print("{:=^34}".format(" ใบเสร็จรับเงิน "))
print("{:<14}{:>6}{:>14}".format("รายการ", "จำนวน", "ราคา"))
print("-" * 34)
total = 0
for name, qty, price in items:
    amount = qty * price
    total += amount
    print("{:<14}{:>6d}{:>14,.2f}".format(name, qty, amount))
print("-" * 34)
print("{:<20}{:>14,.2f}".format("รวมทั้งสิ้น", total))
print("=" * 34)
````

> ⚠️ อักษรไทยที่มีสระบน–ล่างหรือวรรณยุกต์ (เช่น ซ้ำ ที่) ถูกนับเป็นอักขระ แต่ไม่กินพื้นที่บนหน้าจอ คอลัมน์ภาษาไทยจึงอาจเหลื่อมกันเล็กน้อย ส่วนข้อความภาษาอังกฤษและตัวเลขจะตรงคอลัมน์เสมอ

### 🆕 เพิ่มเติม: f-string (แนะนำสำหรับไพธอน 3.6 ขึ้นไป)
**f-string** (formatted string literal) คือวิธีจัดรูปแบบที่สั้นและอ่านง่ายกว่า `format()` โดยใส่ตัวอักษร `f` หน้าสตริง แล้ว **เขียนชื่อตัวแปรหรือนิพจน์ลงใน `{}` ได้โดยตรง** และใช้รหัสรูปแบบหลัง `:` ได้เหมือน `format()` ทุกประการ ปัจจุบันเป็นวิธีที่นิยมที่สุด และ Notebook บทอื่น ๆ ในรายวิชานี้ก็ใช้ f-string เป็นหลัก
</div>

````python
name = "Sammy"
balloons = 5
pi = 3.14159

print("{} has {} balloons.".format(name, balloons))   # แบบ format()
print(f"{name} has {balloons} balloons.")             # แบบ f-string (สั้นกว่า)
print(f"pi ≈ {pi:.2f}")                               # ใช้รหัสรูปแบบได้
print(f"{name:*^20}")                                  # จัดตำแหน่งได้
print(f"5 x 7 = {5 * 7}")                             # ใส่นิพจน์ได้
print(f"{balloons = }")                               # แสดงทั้งชื่อและค่า (สะดวกตอนดีบัก, 3.8+)
````

- `ans1` : ข้อความ `"ราคา 1,250.50 บาท"` จากตัวแปร `price = 1250.5`
- `ans2` : ข้อความ `"รหัสนักศึกษา: 00042"` จากตัวแปร `sid = 42` (เติม 0 ด้านหน้าให้ครบ 5 หลัก)
- `ans3` : ข้อความ `"----Python----"` จากตัวแปร `lang = "Python"` (ฟิลด์กว้าง 14 ตัวอักษร เติม `-` จัดกึ่งกลาง)
</div>

### 💻 ข้อ 5–8: เขียนโค้ดในเซลล์ด้านล่าง
ข้อ 6–8 ให้เก็บผลลัพธ์ไว้ในตัวแปรตามที่กำหนด (ใช้ `format()` หรือ f-string ก็ได้) แล้วรันเซลล์ตรวจคำตอบ

````python
# ข้อ 5: รับชื่อผู้ใช้งานจากแป้นพิมพ์แล้วแสดงผล
````

````python
pi = 3.14159
words = ("dog", "cat", "bat")

q6_1 = ""   # ข้อ 6.1 -> "pi = 3"
q6_2 = ""   # ข้อ 6.2 -> "pi = 3.14"
q6_3 = ""   # ข้อ 6.3 -> "pi = 3.1416"
q7 = ""     # ข้อ 7
q8 = ""     # ข้อ 8

for q in (q6_1, q6_2, q6_3, q7, q8):
    print(q)
````

````python
# ✅ เซลล์ตรวจคำตอบ ข้อ 6-8
expected = {
    "q6_1": "pi = 3",
    "q6_2": "pi = 3.14",
    "q6_3": "pi = 3.1416",
    "q7": "|dog            |      cat      |            bat|",
    "q8": "|dog*******|***cat****|*******bat|",
}
score = 0
for var, exp in expected.items():
    got = globals().get(var)
    ok = got == exp
    score += ok
    print(f"{'✅' if ok else '❌'} {var} = {got!r}" + ("" if ok else f"\n      ควรได้ {exp!r}"))
print(f"\nผ่าน {score}/{len(expected)} ข้อ")
````

1. **Syntax** คือกฎที่ระบุลำดับของสัญลักษณ์ที่ถูกต้องในการเขียนโปรแกรมตามภาษาที่กำหนด ถ้าผิดจะเกิด SyntaxError และโปรแกรมไม่ถูกรันเลย
2. **Semantics** คือความหมายของคำสั่ง หรือสิ่งที่คำสั่งนั้นทำเมื่อถูกรัน โค้ดอาจถูก Syntax แต่ Semantics ผิดได้ เช่น `10 / 0`
3. **Identifiers** คือชื่อที่ใช้ระบุตัวแปร ฟังก์ชัน คลาส โมดูล หรือวัตถุอื่น ๆ
4. **คำสงวน** คือคำที่ไพธอนสงวนไว้ใช้เป็นคำสั่งของภาษา นำมาตั้งชื่อไม่ได้ ในไพธอน 3.13 มี 35 คำ (ตรวจด้วย `len(keyword.kwlist)` เพราะจำนวนอาจต่างกันตามเวอร์ชัน)

### 🏆 แบบทดสอบท้ายบท (ตรวจคะแนนอัตโนมัติ)
รันเซลล์ด้านล่าง แล้วพิมพ์ตัวเลือก (ก/ข/ค/ง) ของแต่ละข้อ

````python
quiz = [
    ("ชื่อใดเป็น Identifier ที่ถูกต้อง",
     ["ก. 2nd_place", "ข. my-score", "ค. _total_2", "ง. first name"], "ค"),
    ("ข้อใดเป็นคำสงวนในไพธอน 3",
     ["ก. print", "ข. True", "ค. input", "ง. main"], "ข"),
    ("ไพธอนใช้สิ่งใดกำหนดขอบเขตของบล็อกโค้ด",
     ["ก. วงเล็บปีกกา { }", "ข. คำว่า begin/end", "ค. เครื่องหมาย : และการเยื้องหน้า", "ง. เครื่องหมาย ;"], "ค"),
    ("ข้อใดใช้เขียน comment หนึ่งบรรทัด",
     ["ก. //", "ข. #", "ค. /* */", "ง. --"], "ข"),
    ("x = input() แล้วผู้ใช้พิมพ์ 10 ตัวแปร x มีชนิดข้อมูลใด",
     ["ก. int", "ข. float", "ค. str", "ง. bool"], "ค"),
    ("print(\"5\" + \"3\") แสดงผลอะไร",
     ["ก. 8", "ข. 53", "ค. 5 3", "ง. Error"], "ข"),
    ("\"{1} {0}\".format(\"A\", \"B\") ได้ผลลัพธ์ใด",
     ["ก. A B", "ข. B A", "ค. {1} {0}", "ง. Error"], "ข"),
    ("\"{:.2f}\".format(3.14159) ได้ผลลัพธ์ใด",
     ["ก. 3.1", "ข. 3.14", "ค. 3.142", "ง. 3.14159"], "ข"),
    ("สัญลักษณ์ใดใช้จัดข้อความให้อยู่กึ่งกลางฟิลด์",
     ["ก. <", "ข. >", "ค. ^", "ง. ="], "ค"),
    ("คำสั่งใดใช้รันไฟล์ test.py ในโหมดสคริปต์",
     ["ก. run test.py", "ข. python test.py", "ค. open test.py", "ง. import test.py"], "ข"),
]

def run_quiz():
    score = 0
    for i, (q, choices, ans) in enumerate(quiz, 1):
        print(f"\nข้อ {i}. {q}")
        for c in choices:
            print("   ", c)
        reply = input("คำตอบ (ก/ข/ค/ง): ").strip()
        if reply == ans:
            print("   ✅ ถูกต้อง"); score += 1
        else:
            print(f"   ❌ ยังไม่ถูก คำตอบที่ถูกคือ {ans}")
    print(f"\n🎯 คะแนนของคุณ: {score}/{len(quiz)}")
    print("🏆 ยอดเยี่ยม!" if score == len(quiz) else "👍 ดีมาก" if score >= 7 else "📖 ทบทวนเนื้อหาอีกครั้งนะ")

run_quiz()
````
